정규표현식 기능을 사용하기 위해 파이썬의 `re` 모듈을 불러옵니다.


```python
import re
```

`a.c` 패턴을 컴파일해 `a`와 `c` 사이에 임의의 문자 하나가 들어가는 문자열을 찾는 예제입니다. `kkk`에는 해당 패턴이 없으므로 검색 결과가 없습니다.


```python
r = re.compile("a.c")
r.search("kkk")
```

앞에서 만든 `a.c` 패턴으로 `abc`를 검색해 점(`.`)이 문자 하나와 매칭되는지 확인합니다.


```python
r.search("abc")
```

정규표현식을 사용해 특수문자 제거, 숫자 제거, 공백 정리, 한글과 영어만 남기기 같은 기본 텍스트 정제 작업을 한 셀에서 확인합니다.


```python
import re

text = "Hello!!! 자연어처리?! 2026년 공부중"

# 1. 특수문자 제거
re.sub(r"[^가-힣a-zA-Z0-9\s]", "", text)    

# 2. 숫자 제거
re.sub(r"\d+", "", text)

# 3. 여러 공백 하나로 줄이기
cleaned = re.sub(r"\s+", " ", text) 
cleaned.strip()

# 4. 한글과 영어만 남기기
re.sub(r"[^가-힣a-zA-Z\s]", "", text)
```

`ab?c` 패턴에서 `?`는 앞의 문자 `b`가 0번 또는 1번 등장하는 경우를 의미합니다. `abbc`는 `b`가 두 번 있으므로 매칭되지 않습니다.


```python
r = re.compile("ab?c")
r.search("abbc")
```

`ab?c` 패턴으로 `abc`를 검색해 `b`가 한 번 있는 문자열이 매칭되는지 확인합니다.


```python
r.search("abc")
```

`ab?c` 패턴으로 `ac`를 검색해 `b`가 없어도 매칭되는지 확인합니다.


```python
r.search("ac")
```

`ab*c` 패턴에서 `*`는 앞의 문자 `b`가 0번 이상 반복되는 경우를 의미합니다. `a`만 있는 문자열은 끝의 `c`가 없어 매칭되지 않습니다.


```python
r = re.compile("ab*c")
r.search("a")
```

`ab*c` 패턴으로 `ac`를 검색해 `b`가 없어도 매칭되는지 확인합니다.


```python
r.search("ac")
```

`ab*c` 패턴으로 `abc`를 검색해 `b`가 한 번 있는 경우도 매칭되는지 확인합니다.


```python
r.search("abc")
```

`ab*c` 패턴으로 `abbbbc`를 검색해 `b`가 여러 번 반복되어도 매칭되는지 확인합니다.


```python
r.search("abbbbc")
```

`ab+c` 패턴에서 `+`는 앞의 문자 `b`가 1번 이상 반복되는 경우를 의미합니다. `ac`는 `b`가 없으므로 매칭되지 않습니다.


```python
r = re.compile("ab+c")
r.search("ac")
```

`ab+c` 패턴으로 `abc`를 검색해 `b`가 한 번 있는 문자열이 매칭되는지 확인합니다.


```python
r.search("abc")
```

`ab+c` 패턴으로 `abbbbc`를 검색해 `b`가 여러 번 반복되는 문자열도 매칭되는지 확인합니다.


```python
r.search("abbbbc")
```

`^ab` 패턴에서 `^`는 문자열의 시작을 의미합니다. 문자열이 `ab`로 시작하는 경우만 찾을 수 있는지 확인합니다.


```python
r = re.compile("^ab")
r.search("bbc")
r.search("zab")
```

`^ab` 패턴으로 `abz`를 검색해 문자열 시작 부분이 `ab`이면 매칭되는지 확인합니다.


```python
r.search("abz")
```

`ab{2}c` 패턴에서 `{2}`는 앞의 문자 `b`가 정확히 두 번 반복되어야 함을 의미합니다. 조건에 맞지 않는 예시들을 먼저 확인합니다.


```python
r = re.compile("ab{2}c")
r.search("ac")
r.search("abc")
r.search("abbbbbc")
```

`ab{2}c` 패턴으로 `abbc`를 검색해 `b`가 정확히 두 번 있는 경우만 매칭되는지 확인합니다.


```python
r.search("abbc")
```

`ab{2,8}c` 패턴에서 `{2,8}`은 `b`가 2번 이상 8번 이하 반복되어야 함을 의미합니다. 범위를 벗어난 예시를 확인합니다.


```python
r = re.compile("ab{2,8}c")
r.search("ac")
r.search("abc")
r.search("abbbbbbbbbc")
```

`ab{2,8}c` 패턴으로 `abbc`를 검색해 반복 횟수가 최소 조건을 만족하는지 확인합니다.


```python
r.search("abbc")
```

`ab{2,8}c` 패턴으로 `abbbbbbbbc`를 검색해 반복 횟수가 최대 조건까지 허용되는지 확인합니다.


```python
r.search("abbbbbbbbc")
```

`a{2,}bc` 패턴에서 `{2,}`는 `a`가 2번 이상 반복되어야 함을 의미합니다. 조건을 만족하지 않는 예시를 먼저 확인합니다.


```python
r = re.compile("a{2,}bc")

r.search("bc")
r.search("aa")
```

`a{2,}bc` 패턴으로 `aabc`를 검색해 `a`가 정확히 두 번 있는 경우 매칭되는지 확인합니다.


```python
r.search("aabc")
```

`a{2,}bc` 패턴으로 `aaaaaaaabc`를 검색해 `a`가 여러 번 반복되어도 매칭되는지 확인합니다.


```python
r.search("aaaaaaaabc")
```

`[abc]` 패턴은 대괄호 안의 문자 중 하나를 찾습니다. `zzz`에는 `a`, `b`, `c`가 없으므로 매칭되지 않습니다.


```python
r = re.compile("[abc]")
r.search("zzz")
```

`[abc]` 패턴으로 `a`를 검색해 대괄호 안 문자 하나가 매칭되는지 확인합니다.


```python
r.search("a")
```

`[abc]` 패턴으로 `aaaaaa`를 검색해 반복된 문자열에서도 첫 매칭 위치를 찾는지 확인합니다.


```python
r.search("aaaaaa")
```

`[abc]` 패턴으로 `baac`를 검색해 문자열 안에서 대괄호에 포함된 문자를 찾는지 확인합니다.


```python
r.search("baac")
```

`[a-z]` 패턴은 소문자 알파벳 범위를 찾습니다. 대문자와 숫자에는 매칭되지 않는지 확인합니다.


```python
r = re.compile("[a-z]")

r.search("AAA")
r.search("111")
```

`[a-z]` 패턴으로 `aBC`를 검색해 소문자가 포함된 경우 매칭되는지 확인합니다.


```python
r.search("aBC")
```

`[^abc]` 패턴에서 `^`는 대괄호 안에서 제외를 의미합니다. `a`, `b`, `c`만 있는 문자열에는 매칭되지 않는지 확인합니다.


```python
r = re.compile("[^abc]")
r.search("a")
r.search("ab")
r.search("b")
```

`[^abc]` 패턴으로 `d`를 검색해 제외 목록에 없는 문자가 매칭되는지 확인합니다.


```python
r.search("d")
```

`[^abc]` 패턴으로 `1`을 검색해 숫자도 제외 목록에 없으면 매칭되는지 확인합니다.


```python
r.search("1")
```

`match()`는 문자열의 시작부터 패턴을 검사합니다. `ab.` 패턴이 중간에 있는 경우 `match()`로는 찾지 못하는지 확인합니다.


```python
r = re.compile("ab.")
r.match("kkkabc")
```

`search()`는 문자열 전체에서 패턴을 찾습니다. 같은 `ab.` 패턴이 중간에 있어도 검색되는지 확인합니다.


```python
r.search("kkkabc")
```

`match()`로 `abckkk`를 검사해 문자열 시작 부분이 패턴과 맞으면 매칭되는지 확인합니다.


```python
r.match("abckkk")
```

공백 한 칸을 기준으로 과일 이름 문자열을 나누어 `re.split()`의 기본 사용법을 확인합니다.


```python
text = "사과 딸기 수박 메론 바나나"
re.split(" ", text)
```

줄바꿈 문자(`
`)를 기준으로 여러 줄 문자열을 나누는 예제입니다.


```python
text = """사과
딸기
수박
메론
바나나"""
re.split("\n", text)
```

더하기 기호(`+`)를 구분자로 사용해 문자열을 나눕니다. `+`는 정규표현식 특수문자이므로 역슬래시로 이스케이프합니다.


```python
text = "사과+딸기+수박+메론+바나나"
re.split("\+", text)
```

`findall()`과 `\d+` 패턴을 사용해 문장 안의 숫자 묶음을 모두 추출합니다.


```python
text = """이름 : 김철수
전화번호 : 010 - 1234 - 1234
나이 : 30
성별 : 남"""
re.findall("\d+", text)
```

숫자가 없는 문자열에 `\d+` 패턴을 적용하면 빈 리스트가 반환되는지 확인합니다.


```python
re.findall("\d+", "문자열입니다.")
```

번호, 이름, 직무 코드가 섞여 있는 예제 문자열을 만들어 이후 정규표현식 추출 실습에 사용합니다.


```python
text = """100 John    PROF
101 James   STUD
102 Mac   STUD"""
```

`\s+` 패턴으로 하나 이상의 공백을 기준으로 문자열을 나누어 불규칙한 공백을 처리합니다.


```python
re.split("\s+", text)
```

`\d+` 패턴으로 예제 문자열에서 숫자 ID만 추출합니다.


```python
re.findall('\d+', text)
```

`[A-Z]` 패턴으로 예제 문자열에서 대문자 한 글자씩 추출합니다.


```python
re.findall('[A-Z]', text)
```

`[A-Z]{4}` 패턴으로 대문자 네 글자가 연속된 직무 코드만 추출합니다.


```python
re.findall('[A-Z]{4}', text)
```

`[A-Z][a-z]+` 패턴으로 첫 글자는 대문자, 뒤에는 소문자인 영어 이름을 추출합니다.


```python
re.findall('[A-Z][a-z]+', text)
```

NLTK의 `RegexpTokenizer`를 사용해 단어 문자 기준 토큰화와 공백 기준 토큰화를 비교합니다.


```python
from nltk.tokenize import RegexpTokenizer

text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop"

tokenizer1 = RegexpTokenizer("[\w]+")
tokenizer2 = RegexpTokenizer("\s+", gaps=True)

print(tokenizer1.tokenize(text))
print(tokenizer2.tokenize(text))
```
